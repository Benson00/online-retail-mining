```python
import os
import math
from PIL import Image

def combine_png_images(folder_name, output_name="dashboard_custom.png", max_cols=2):
    png_files = [f for f in os.listdir(folder_name) if f.lower().endswith('.png')]
    png_files.sort()
    
    if not png_files:
        return

    image_paths = [os.path.join(folder_name, f) for f in png_files]
    images = [Image.open(p) for p in image_paths]

    num_images = len(images)
    cols = min(max_cols, num_images)
    rows = math.ceil(num_images / cols)

    max_width = max(img.width for img in images)
    max_height = max(img.height for img in images)

    canvas_width = cols * max_width
    canvas_height = rows * max_height

    final_image = Image.new('RGBA', (canvas_width, canvas_height), (255, 255, 255, 255))

    for index, img in enumerate(images):
        row = index // cols
        col = index % cols
        
        images_in_this_row = min(cols, num_images - row * cols)
        row_start_x = (cols - images_in_this_row) * max_width // 2
        
        x_offset = row_start_x + (col * max_width)
        y_offset = row * max_height
        
        final_image.paste(img, (x_offset, y_offset))

    final_image.save(output_name)




folders = ["results10core", "results20core", "results4core", "scalability"]

for folder in folders:
    output_file = f"dashboard_{folder}.png"
    combine_png_images(folder_name=folder, output_name=output_file, max_cols=2)
```
